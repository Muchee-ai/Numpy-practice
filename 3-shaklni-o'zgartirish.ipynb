Massivlar pozitsiyasini ko'chirish(Transpose), o'qlar o'rnini almashtirish(swapping axis)

```python
a = np.arange(6).reshape(3,2) #reshape metodi orqali
a
```

```python
a.T #a massivning transpose ko'rinishi
```

```python
np.dot(a, a.T)
```

```python
a.reshape(2, -1)
```
