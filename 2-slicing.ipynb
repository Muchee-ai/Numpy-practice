Boolean indekslash

```python
names = np.array(['Hasan', 'Husan', 'Otabek', 'Dilmurod', 'Husan', 'Javohir'])
names
```

```python
data = np.random.randn(6,4)
data
```

```python
names == 'Hasan'
```

```python
data[names == 'Hasan'] #Hasanga tegishli barcha ma'lumotlar
```

```python
data[names == 'Husan'] #Husanga tegishli barcha ma'lumotlar
```

```python
data[names == 'Javohir', 2:] #Javohga tegishli oxrgi 2ta ustundagi ma'lumotlar
```

```python
data[names != 'Hasan'] #Hasandan tashqari barcha insonlarni ma'lumotlari
```

```python
data[~(names == 'Hasan')] #(~)yuqoridagi cell bn brxil natija
```

```python
print(names)
print(data)
```

```python
mask = (names == 'Hasan') | (names == 'Husan') #hasan/husanga tegishli ma'lumotlar
mask
```

```python
data[mask]
```
