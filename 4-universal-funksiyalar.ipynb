```python
import numpy as np
```

Unary funksiyalar (bitta argument qabul qiladi)
[sqrt, square, exp, log, modf, sign, isnan]

```python
arr = np.arange(10)
arr
```

```python
np.sqrt(arr) #massivning harbir elementini ildizni qaytaradi
```

```python
np.square(arr) #masssivning harbir elementini kvadratga oshiradi
```

```python
np.exp(arr) #elementlarni eksponentini qaytaradi
```

```python
np.log(arr[1:]) #massivning harbir elementini logarifmini qaytaradi
```

modf - haqiqy sonlarni butun va qoldiq qismlarga ajratib beruvchi funksiya


```python
arr = np.random.randn(6)
arr
```

```python
butun, qoldiq = np.modf(arr)
print(butun)
print(qoldiq)
```

```python
np.isnan(arr) #massivdagi qaysi elementlar NaN ekanini tekshiradi.
```

```python
np.sign(arr) #sonning ishorasini qaytaradi. Ya'ni son musbatmi, manfiymi yoki nolmi.
```

Binary funksiyalar(ikkita argument qabul qiladi) (add, multiple, maximum)

```python
arr1 = np.random.randn(6)
arr2 = np.random.randn(6)
```

```python
arr1
```

```python
arr2
```

```python
np.add(arr1, arr2) ##arr1 va arr2 elementlarini mos ravishda qo'shib beradi
```

```python
arr1+arr2
```

```python
np.multiply(arr1, arr2) ##arr1 va arr2 elementlarini mos ravishda ko'paytirib beradi
```

```python
np.maximum(arr1, arr2) ##ikki elementlarni taqqoslab, eng kattasini qaytaradi
```
